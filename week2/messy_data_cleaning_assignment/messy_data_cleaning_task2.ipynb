```python
import pandas as pd
```

```python
df = pd.read_csv('online_retail_real_world.csv')
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df['ProductName'].value_counts()
```

```python
df['Brand'].value_counts()
```

```python
df['Brand'] = df['Brand'].fillna('missing')
```

```python
df['Brand'].value_counts()
```

```python
product_name_mode = df['ProductName'].mode()[0]
product_name_mode
```

```python
df['ProductName'] = df['ProductName'].fillna(product_name_mode)
```

```python
df['ProductName'].value_counts()
```

```python
df.isna().sum()
```

```python
df['Raw_Weight'] = df['Raw_Weight'].astype(str).str.replace('g', '' ,case=False).str.strip()
```

```python
df
```

```python
df['Raw_Weight'] = pd.to_numeric(df['Raw_Weight'], errors='coerce')
raw_weight_mean = df['Raw_Weight'].mean()
raw_weight_mean
```

```python
df['Raw_Weight'] = df['Raw_Weight'].fillna(raw_weight_mean)
```

```python
df.isna().sum()
```

```python
UnitPrice_mean = df['UnitPrice'].mean()
UnitPrice_mean
```

```python
df['UnitPrice'] = df['UnitPrice'].fillna(UnitPrice_mean)
```
