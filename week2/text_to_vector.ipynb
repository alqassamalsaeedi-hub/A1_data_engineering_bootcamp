```python
import numpy as np 
import re
```

```python
def text_to_vector(text, bag_of_words):
    text = text.lower()
    words = re.findall(r'\b\w+\b', text)
    vector = np.zeros(len(bag_of_words), dtype = int)
    for i, word in enumerate(bag_of_words):
        vector[i] = words.count(word.lower())
    return vector
```

### example

```python
bag_of_words= ['hello', 'world']
text = 'hello'
text_to_vector(text, bag_of_words)
```
