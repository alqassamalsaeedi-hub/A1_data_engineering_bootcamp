#  Multiple Inheritance
* it's the ability for a class to inherite the properties and methods of multuiple classes in the same time

# example

```python
class Father:
    def skill_a(self):
        print("Driving skills from Father")


class Mother:
    def skill_b(self):
        print("Cooking skills from Mother")


class Child(Father, Mother):
    def own_skill(self):
        print("Gaming skills of the Child")


c = Child()
c.skill_a()  
c.skill_b()  
c.own_skill() 
```
