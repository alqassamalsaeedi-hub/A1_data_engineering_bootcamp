```python
import requests
from bs4 import BeautifulSoup
import os 
import pandas

url = "https://www.x-rates.com/table/?from=USD&amount=1"
headers = {'User-Agent': 'Mozilla/5.0'}

response = requests.get(url, headers= headers)
soup = BeautifulSoup(response.text, 'html.parser')
print(soup)

```

```python
tables = soup.find_all('td')
```

```python
tables
```

```python
for i in tables:
    print(i.get_text())
```

```python
soup.find_all('table')
```

```python
a = soup.find("table")
print(a.get_text())
```

```python
a = soup.find("tr")
print(a.get_text())
```

```python
a = soup.find('table', class_='tablesorter')
print(a.get_text())
```

```python
rows = a.find_all('tr')[1:16]
for i in rows:
    print(i.get_text())
```
